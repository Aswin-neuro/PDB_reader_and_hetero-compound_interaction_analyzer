- a. List of atoms of any residue speciﬁed by residue number; listAtoms(i)
- b. Atomic coordinates of any atom speciﬁed by its residue and atom name;
coordAtom(i,’atomName’).
- c. All heterocompunds (Identiﬁer) in the PDB; hetero()
- d. Distance between two atoms; distanceAtoms(i,’atomName_i’,j,’atomName_j)
Using the above object and methods ﬁnd hetero compound having the maximum average
atomic interactions. For this, ﬁrst ﬁnd number of interacting atoms for each hetero-
compound and divide it by the number of atoms of hetero compound.
- Two atoms are said to be interacting if distance between two atoms is within 5 Å.

````python
pdb=(open('proj5.pdb','r'))

atom_data={}
Hetdict={}
for line in pdb:
  # Extracting from lines

  H_A=line[0:6].strip()
  res_no=int(line[22:26].strip())
  atom=line[12:16].strip()
  res_name=line[17:20].strip()
  xcord=line[30:38].strip()
  ycord=line[38:46].strip()
  zcord=line[46:54].strip()
  # Making dicts
  if H_A != 'TER':
    if res_no not in atom_data:
      atom_data[res_no]={
          'res_name': res_name,
          'H_A':H_A,
          atom:[xcord,ycord,zcord]
      }
    else:
      atom_data[res_no][atom]=[xcord,ycord,zcord]
    if H_A=='HETATM':
      if res_no not in Hetdict:
        Hetdict[res_no]={'res_name': res_name, atom:[xcord,ycord,zcord]}
      else:
        Hetdict[res_no][atom]=[xcord,ycord,zcord]
pdb.close()
#Class

class PDB_Reader:
  def __init__(self,info_dict):
    self.info_dict=info_dict

  def listAtoms(self,i):
    atom_data=self.info_dict
    atomlst=[]
    for key in atom_data[i].keys():
      if key!='res_name'and key!='H_A':
        atomlst.append(key)
    print(f'The list of atoms for {i}th residue is{atomlst}',end='')

  def coordAtom(self,i,name):
    atom_data=self.info_dict
    print(atom_data[i][name])

  def Hetero(self):
    atom_data=self.info_dict
    Hetero_dict={}
    for res_no in atom_data.keys():
      for H_A in atom_data[res_no].keys():
        if H_A=='H_A':
          if atom_data[res_no][H_A]=='HETATM':
            print(atom_data[res_no]['res_name'])

  def distanceAtoms(self,i1,i2,atom_n1,atom_n2):
    atom_data=self.info_dict
    x1=float(atom_data[i1][atom_n1][0])
    y1=float(atom_data[i1][atom_n1][1])
    z1=float(atom_data[i1][atom_n1][2])
    x2=float(atom_data[i2][atom_n2][0])
    y2=float(atom_data[i2][atom_n2][1])
    z2=float(atom_data[i2][atom_n2][2])
    import math as mt
    xdiff=(x1-x2)**2
    ydiff=(y1-y2)**2
    zdiff=(z1-z2)**2
    result = mt.sqrt(xdiff+ydiff+zdiff)
    return result
# Finding interactions
value_list=[]
def main():
  x=PDB_Reader(atom_data)
  for id1 in Hetdict.keys():
    count=0
    for key1 in Hetdict[id1].keys():
      atomcnt=0
      if key1!='res_name':
        for id2 in atom_data.keys():
          for key2 in atom_data[id2].keys():
            if key2!='res_name' and key2!='H_A':
              values=(x.distanceAtoms(id1,id2,key1,key2))
              if 0<values<5:
                count+=1

    avgint=(count)/(len(Hetdict[id1].keys())-1)
    print(f'for {id1} the average interaction is {avgint}')
  x=atom_data[546]
  print(f'The heteroatom with the highest interaction is {x}')

# Class PDB_Reader Usage
def PDB_usage():
  x=PDB_Reader(atom_data)
  # listAtom - usage is inputting any number for the details of the residue you want
  x.listAtoms(5)

  # coordAtom - usage is inputting number of residue 'i' and name of atom eg: 'O'
  # residue_number=int(in)
  x.coordAtom(5,'C')

  #Hetero will print all the heteroatoms
  hetero=x.Hetero()

  #distanceAtoms needs 4 variables - (i,j,atomname of ith residue,atomname of jth residue) i and j are numbers from 4 to 549 and atomname should for the same can be found from listAtoms(i) or (j)
  distance=x.distanceAtoms(4,5,'CA','N')
  print(f'the distance between the mentioned atoms is {distance}')

main()
PDB_usage()
````
